```python
df2 = df.copy()
```

```python
df2['Age'] = df2['Age'].fillna(df2['Age'].mean())
```

```python
df2['Embarked'] = df2['Embarked'].fillna(df2['Embarked'].mode()[0])
```

```python
df2['Cabin'] = df2['Cabin'].fillna('Unknown')
```

```python
df2.isnull().sum()
```

```python
df2['Sex'] = df2['Sex'].map({'male': 1, 'female': 0})
```

```python
df2.corr(numeric_only=True)
```

```python
df2.to_csv('Titanic_Cleaned.csv', index=False)
```
